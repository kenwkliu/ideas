# **Build Association Rules**

Convert itemsets into rules and caluculate confidence and lift to find out reliability of the associated brands, and whether the reliability is better than expected by chance.

```
import pandas as pd
from mlxtend.frequent_patterns import association_rules

url = "https://raw.githubusercontent.com/kenwkliu/ideas/master/colab/data/beautyFanPages.csv"
df = pd.read_csv(url).fillna(0)

brands = [
    "LaneigeHongKong", "HKinnisfree", "TheBodyShopHongKong",
    "Bioderma.HongKong", "yslbeaute.hk", "CliniqueHongKong",
    "etude.hk", "shuuemura.hk", "armanibeautyHK",
    "esteelauderhk", "loccitane.hk", "SHISEIDOHK"
]

transactions = df[brands].astype(bool)

frequent_itemsets = apriori(
    transactions,
    min_support=0.001,
    use_colnames=True
)

rules = association_rules(
    frequent_itemsets,
    metric="confidence",
    min_threshold=0.10
)

rules = rules.sort_values(
    by=["lift", "confidence", "support"],
    ascending=False
)

rules[[
    "antecedents", "consequents",
    "support", "confidence", "lift"
]].head(20)
```

```
#@title Choose a particular company for its associated companies

company = "SHISEIDOHK"  #@param [
# "LaneigeHongKong",
# "HKinnisfree",
# "TheBodyShopHongKong",
# "Bioderma.HongKong",
# "yslbeaute.hk",
# "CliniqueHongKong",
# "etude.hk",
# "shuuemura.hk",
# "armanibeautyHK",
# "esteelauderhk",
# "loccitane.hk",
# "SHISEIDOHK"
# ]

company_rules = rules[
    rules["antecedents"].apply(lambda items: company in items)
].sort_values(
    by=["lift", "confidence", "support"],
    ascending=False
)

display(
    company_rules[
        [
            "antecedents", "consequents",
            "support", "confidence", "lift"
        ]
    ].head(20)
)
```

# **Create Reliability Rules**
Take the analysis further by setting support, confidence and lift levels to find out reliability of the associated brands.

**1. Support**

Brand co-occurrence among the whole dataset.

**2. Confidence**

Among users who interacted with the antecedent brands, the proportion who also interacted with the consequent brands.

**3. Lift**

Compare the observed co-occurrence with what would be expected if the two sides were unrelated.

Lift > 1 indicates positive association.

The higher the Lift, the stronger the association relative to the consequent's general popularity.

```
rules = association_rules(
    frequent_itemsets,
    metric="confidence",
    min_threshold=0.10
)

# Thresholds
MIN_SUPPORT = 0.001
MIN_CONFIDENCE = 0.10
MIN_LIFT = 1.20

reliable_rules = rules[
    (rules["support"] >= MIN_SUPPORT) &
    (rules["confidence"] >= MIN_CONFIDENCE) &
    (rules["lift"] >= MIN_LIFT)
].copy()

reliable_rules["antecedent_brand"] = reliable_rules["antecedents"].apply(
    lambda x: ", ".join(sorted(list(x)))
)

reliable_rules["consequent_brand"] = reliable_rules["consequents"].apply(
    lambda x: ", ".join(sorted(list(x)))
)

reliable_rules = reliable_rules.sort_values(
    by=["lift", "confidence", "support"],
    ascending=[False, False, False]
)

reliable_rules[[
    "antecedent_brand",
    "consequent_brand",
    "support",
    "confidence",
    "lift"
]].head(20)
```

```
# Rank individual brands by association_score

antecedent_scores = (
    reliable_rules
    .assign(association_score=reliable_rules["lift"] * reliable_rules["support"])
    .groupby("antecedent_brand")
    .agg(
        rules_as_antecedent=("antecedent_brand", "size"),
        mean_confidence=("confidence", "mean"),
        mean_lift=("lift", "mean"),
        total_support=("support", "sum"),
        association_score=("association_score", "sum")
    )
    .sort_values(
        by=["association_score", "mean_lift"],
        ascending=False
    )
)

antecedent_scores.head(12)
```

```
from google.colab import files


reliable_rules = reliable_rules.copy()
reliable_rules["association_score"] = (
    reliable_rules["lift"] * reliable_rules["support"]
)

outgoing_scores = (
    reliable_rules
    .groupby("antecedent_brand", as_index=False)
    .agg(
        rules_as_antecedent=("antecedent_brand", "size"),
        avg_confidence=("confidence", "mean"),
        avg_lift=("lift", "mean"),
        total_rule_support=("support", "sum"),
        association_score=("association_score", "sum")
    )
    .rename(columns={"antecedent_brand": "brand"})
)

incoming_scores = (
    reliable_rules
    .groupby("consequent_brand", as_index=False)
    .agg(
        rules_as_consequent=("consequent_brand", "size"),
        avg_confidence_received=("confidence", "mean"),
        avg_lift_received=("lift", "mean"),
        total_rule_support_received=("support", "sum"),
        association_score_received=("association_score", "sum")
    )
    .rename(columns={"consequent_brand": "brand"})
)

brand_ranking = outgoing_scores.merge(
    incoming_scores,
    on="brand",
    how="outer"
).fillna(0)


brand_ranking["total_association_score"] = (
    brand_ranking["association_score"] +
    brand_ranking["association_score_received"]
)

brand_ranking = brand_ranking.sort_values(
    by=[
        "total_association_score",
        "avg_lift",
        "avg_confidence"
    ],
    ascending=False
).reset_index(drop=True)

brand_ranking.insert(
    0,
    "rank",
    range(1, len(brand_ranking) + 1)
)

brand_ranking = brand_ranking.round(4)


rules_output = reliable_rules[[
    "antecedent_brand",
    "consequent_brand",
    "support",
    "confidence",
    "lift",
    "association_score"
]].sort_values(
    by=["association_score", "lift", "confidence", "support"],
    ascending=False
).round(4)


brand_ranking.to_csv(
    "brand_association_ranking.csv",
    index=False
)

rules_output.to_csv(
    "reliable_brand_rules.csv",
    index=False
)

display(brand_ranking)

files.download("brand_association_ranking.csv")
files.download("reliable_brand_rules.csv")
```

```
# Clean holding ticker symbols
# Request Yahoo holdings
# Remove invalid or duplicate holding symbols
# Rank by weight, keeps max 10
# Exclude funds with <8 holdings

def clean_ticker(value):
    if pd.isna(value):
        return None

    value = str(value).strip().upper()

    invalid_values = {
        "",
        "NAN",
        "NONE",
        "-",
        "CASH",
        "N/A"
    }

    if value in invalid_values:
        return None

    return value


def get_top_holdings(fund_ticker):
    try:
        raw = yf.Ticker(fund_ticker).funds_data.top_holdings

        if raw is None or raw.empty:
            return pd.DataFrame()

        holdings = raw.reset_index()

        columns_lower = {
            str(column).lower().strip(): column
            for column in holdings.columns
        }

        symbol_col = next(
            (
                original for lower, original in columns_lower.items()
                if lower in {"symbol", "ticker"}
                or "symbol" in lower
                or "ticker" in lower
            ),
            None
        )

        name_col = next(
            (
                original for lower, original in columns_lower.items()
                if lower == "name"
                or "holding name" in lower
                or "company" in lower
            ),
            None
        )

        weight_col = next(
            (
                original for lower, original in columns_lower.items()
                if "holding percent" in lower
                or "weight" in lower
                or "percent" in lower
            ),
            None
        )

        if symbol_col is None:
            return pd.DataFrame()

        output = pd.DataFrame({
            "holding_ticker": holdings[symbol_col].map(clean_ticker)
        })

        output["holding_name"] = (
            holdings[name_col]
            .fillna("")
            .astype(str)
            .str.strip()
            if name_col is not None
            else output["holding_ticker"]
        )

        output["holding_weight"] = (
            pd.to_numeric(holdings[weight_col], errors="coerce")
            if weight_col is not None
            else np.nan
        )

        output = (
            output
            .dropna(subset=["holding_ticker"])
            .drop_duplicates(subset=["holding_ticker"])
        )

        if output["holding_weight"].notna().any():
            output = output.sort_values(
                "holding_weight",
                ascending=False
            )

        output = output.head(MAX_HOLDINGS).reset_index(drop=True)

        if len(output) < MIN_HOLDINGS:
            return pd.DataFrame()

        output.insert(0, "fund_name", fund_ticker)
        output.insert(1, "holding_rank", range(1, len(output) + 1))

        return output

    except Exception:
        return pd.DataFrame()
```

#**Run Apriori**

 Build market basket, run apriori, generates directional rules, filters reliable rules, and rank the most associated holdings, find crowded stocks, crowded pairs, and rare holdings.

```
# Build one market basket per fund

market_basket = (
    holdings.assign(held=True)
    .pivot_table(
        index="fund_name",
        columns="holding_ticker",
        values="held",
        aggfunc="max",
        fill_value=False
    )
    .astype(bool)
)

print("Market-basket shape:", market_basket.shape)
print("Rows = funds:", market_basket.shape[0])
print("Columns = unique holding tickers:", market_basket.shape[1])

display(market_basket.iloc[:10, :20])
```

```
# Set rules and run apriori

MIN_SUPPORT = 0.02
MAX_ITEMSET_LENGTH = 3

frequent_itemsets = apriori(
    market_basket,
    min_support=MIN_SUPPORT,
    use_colnames=True,
    max_len=MAX_ITEMSET_LENGTH,
    low_memory=True
)

frequent_itemsets["itemset_size"] = (
    frequent_itemsets["itemsets"]
    .apply(len)
)

frequent_itemsets["itemset_label"] = (
    frequent_itemsets["itemsets"]
    .apply(lambda items: " + ".join(sorted(items)))
)

frequent_itemsets = (
    frequent_itemsets
    .sort_values(
        ["itemset_size", "support"],
        ascending=[True, False]
    )
    .reset_index(drop=True)
)

print("Frequent itemsets found:", len(frequent_itemsets))
display(frequent_itemsets.head(30))
```

```
# Directional rules

MIN_CONFIDENCE = 0.30

rules = association_rules(
    frequent_itemsets,
    metric="confidence",
    min_threshold=MIN_CONFIDENCE
)

rules["antecedent_label"] = (
    rules["antecedents"]
    .apply(lambda items: " + ".join(sorted(items)))
)

rules["consequent_label"] = (
    rules["consequents"]
    .apply(lambda items: " + ".join(sorted(items)))
)

rules["rule"] = (
    rules["antecedent_label"] +
    " -> " +
    rules["consequent_label"]
)

rules["antecedent_size"] = rules["antecedents"].apply(len)
rules["consequent_size"] = rules["consequents"].apply(len)

rules = (
    rules
    .sort_values(
        ["lift", "confidence", "support"],
        ascending=False
    )
    .reset_index(drop=True)
)

print("Directional rules generated:", len(rules))

display(
    rules[
        [
            "rule",
            "support",
            "confidence",
            "lift",
            "leverage",
            "conviction"
        ]
    ].head(30)
)
```

```
# Rank associations

RELIABLE_MIN_SUPPORT = 0.02
RELIABLE_MIN_CONFIDENCE = 0.40
RELIABLE_MIN_LIFT = 1.20

reliable_rules = (
    rules.loc[
        (rules["support"] >= RELIABLE_MIN_SUPPORT) &
        (rules["confidence"] >= RELIABLE_MIN_CONFIDENCE) &
        (rules["lift"] >= RELIABLE_MIN_LIFT)
    ]
    .copy()
    .sort_values(
        ["lift", "confidence", "support"],
        ascending=False
    )
    .reset_index(drop=True)
)

print("Reliable directional rules:", len(reliable_rules))

display(
    reliable_rules[
        [
            "rule",
            "support",
            "confidence",
            "lift",
            "leverage",
            "conviction"
        ]
    ].head(30)
)
```

#**Findings**

Apriori can be used as a screening and overlap-detection tool over 4,613 funds to find out:

1. The individual stocks held by many funds
2. Pairs and groups of holdings that frequently appear together
3. Directional relationships
4. Rare holdings that appear in no more than 1% of funds

```
#  Identify crowded stocks, crowded pairs, and rare holdings

N_FUNDS = market_basket.shape[0]

# Individual ticker popularity
ticker_summary = pd.DataFrame({
    "holding_ticker": market_basket.columns,
    "fund_count": market_basket.sum(axis=0).astype(int).values
})

ticker_summary["support"] = (
    ticker_summary["fund_count"] / N_FUNDS
)

ticker_summary = ticker_summary.sort_values(
    ["fund_count", "holding_ticker"],
    ascending=[False, True]
).reset_index(drop=True)

crowded_stocks = ticker_summary.copy()

# Common pairs: select only 2-ticker frequent itemsets
crowded_pairs = (
    frequent_itemsets.loc[
        frequent_itemsets["itemset_size"] == 2,
        ["support", "itemsets", "itemset_label"]
    ]
    .copy()
)

crowded_pairs["fund_count"] = (
    crowded_pairs["support"] * N_FUNDS
).round().astype(int)

crowded_pairs = crowded_pairs.sort_values(
    ["fund_count", "support"],
    ascending=False
).reset_index(drop=True)

# Rare means held by 1% or fewer of included funds
RARE_MAX_SUPPORT = 0.01

rare_holdings = (
    ticker_summary.loc[
        ticker_summary["support"] <= RARE_MAX_SUPPORT
    ]
    .sort_values(
        ["fund_count", "holding_ticker"],
        ascending=[True, True]
    )
    .reset_index(drop=True)
)

print("\nMost crowded stocks:")
display(crowded_stocks.head(30))

print("\nMost crowded ticker pairs:")
display(crowded_pairs.head(30))

print("\nRare holdings:")
display(rare_holdings.head(30))
```

**Niche pattern**

These rules had about 2.1% – 2.3% support, so only around 95 – 108 of the 4,613 funds showed the combination. But their confidence was  85% – 100%, meaning the holdings nearly always appeared together within that smaller group.


**Widespread pattern**
* Microsoft (MSFT): held by about 32.4% of funds.
* NVIDIA (NVDA): about 26.5%.
* Amazon (AMZN): about 26.3%.
* Apple (AAPL): about 25.9%.
* Broadcom (AVGO): about 24.2%.
* Alphabet shares (GOOGL and GOOG): each appear frequently, though separately counted as different share classes.

These are widespread because they are large, well-known companies that can fit several strategies, whether it’s broad US stocks. equity funds, growth funds, technology funds, or index-tracking funds. Fund managers do not need to follow the same provider or detailed model portfolio to hold Microsoft or Apple, many can arrive at the same choice independently.

###**Implications to investors**
A high-confidence association can signal overlap, while high support for mega-cap stocks can signal crowded exposure.

The stocks under niche patterns might suggest hidden duplication that a tightly linked group of fund holdings potentially by a shared provider, model portfolio, or specialised allocation. Buying several funds with that pattern may not diversify the investment as much as their names suggest.

The stocks under widespread patterns suggest concentration risk that buying different funds might result in increased exposure in the same stocks e.g. MSFT appears in almost 1/3 of all the analysed funds.

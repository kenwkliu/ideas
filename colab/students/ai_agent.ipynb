### 4. Extra/ Good to know

#### 4.1 Git
We can initialize a git **bare repository** in google drive and push commit to that upstream.

First we mount the drive.
```python
from google.colab import drive
drive.mount('/content/drive')
```

Next, we change directory to the Google Drive
```bash
cd drive/MyDrive
git init --bare <repo-name>.git

# After Initialized
cd /content # default working dir of colab
git clone /content/drive/MyDrive/<repo-name>.git
```

This requires a good understanding about git. (working dir, local dir, staging, remote tracking branch, stash) For reference: https://stackoverflow.com/questions/37992400/what-is-a-bare-repository-and-why-would-i-need-one


#### 4.2 Web Serve

We can also serve the frontend of opencode through some free ``tunnels``
For example [serveo](https://serveo.net/) or [pinggy](https://pinggy.io/).

The command was like this
```bash
opencode serve
ssh -R 80:localhost:4096 serveo.net
ssh -p 443 -R0:127.0.0.1:4096 free.pinggy.io
```

This requires some general understanding towards networking/tunneling/ssh/proxy.
> Please be reminded that never use ``http`` in the public, as http has no encryption/srambling and the data transmitted can be stolen. Please always use ``https``. (Http can be used within the router, but not in the public) Having certain understanding regarding certificates/openssl is beneficial as we can avoid privacy problems.
